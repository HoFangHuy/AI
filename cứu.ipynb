```python
import os
import cv2
import json
import numpy as np
from ultralytics import YOLO
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, Flatten
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.preprocessing import image
from tensorflow.keras.optimizers import Adam
from google.colab import drive

# 1. Kết nối Google Drive
drive.mount('/content/drive')

# 2. Đường dẫn
base_path = '/content/drive/MyDrive/Data AI'
dataset_path = os.path.join(base_path, 'dataset')
cropped_path = os.path.join(base_path, 'cropped_images')
menu_path = os.path.join(base_path, 'menu.json')
input_image_path = os.path.join(base_path, 'input_tray.jpg')
model_path = os.path.join(base_path, 'cnn_food_model.h5')

# Tạo thư mục nếu chưa tồn tại
os.makedirs(cropped_path, exist_ok=True)
os.makedirs(dataset_path, exist_ok=True)

# 3. Tạo file menu.json
menu = [
    {"name": "thit_kho", "price": 10000},
    {"name": "thit_kho_trung", "price": 10000},
    {"name": "trung_chien", "price": 5000},
    {"name": "ga_chien", "price": 15000}
]
with open(menu_path, 'w') as f:
    json.dump(menu, f, indent=4)

# 4. Tăng cường dữ liệu
def create_data_generator():
    return ImageDataGenerator(
        rescale=1./255,
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True,
        brightness_range=[0.7, 1.3],
        fill_mode='nearest',
        validation_split=0.2
    )

# 5. Tải dữ liệu
def load_data():
    datagen = create_data_generator()
    train_generator = datagen.flow_from_directory(
        dataset_path,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='training'
    )
    val_generator = datagen.flow_from_directory(
        dataset_path,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='validation'
    )
    return train_generator, val_generator

# 6. Xây dựng mô hình CNN với ResNet50
def build_cnn_model(num_classes):
    base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    for layer in base_model.layers[:-10]:
        layer.trainable = False  # Freeze các tầng đầu, fine-tune các tầng cuối

    model = Sequential([
        base_model,
        Flatten(),
        Dense(256, activation='relu'),
        Dropout(0.5),
        Dense(128, activation='relu'),
        Dropout(0.3),
        Dense(num_classes, activation='softmax')
    ])

    model.compile(optimizer=Adam(learning_rate=0.0001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# 7. Huấn luyện mô hình
def train_cnn_model():
    train_generator, val_generator = load_data()
    num_classes = len(train_generator.class_indices)

    if num_classes != 4:
        raise ValueError(f"Dataset phải chứa đúng 4 lớp, tìm thấy {num_classes} lớp!")

    model = build_cnn_model(num_classes)
    history = model.fit(
        train_generator,
        validation_data=val_generator,
        epochs=20,  # Tăng epochs để đạt độ chính xác cao
        verbose=1
    )
    model.save(model_path)
    return model, train_generator.class_indices

# 8. Dùng YOLO để cắt ảnh món ăn
def crop_foods(image_path, output_dir):
    # Load YOLOv8 (giả định đã fine-tune cho món ăn, ở đây dùng cấu hình cơ bản)
    model = YOLO('yolov8n.pt')  # Cần fine-tune trên dataset món ăn
    results = model(image_path)

    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Không thể đọc ảnh từ {image_path}")

    cropped_images = []
    for result in results:
        boxes = result.boxes.xyxy.cpu().numpy()
        for i, box in enumerate(boxes):
            x1, y1, x2, y2 = map(int, box)
            cropped_img = img[y1:y2, x1:x2]
            cropped_img = cv2.resize(cropped_img, (224, 224))
            output_path = os.path.join(output_dir, f'food_{i}.jpg')
            cv2.imwrite(output_path, cropped_img)
            cropped_images.append(output_path)

    return cropped_images

# 9. Nhận diện món ăn và tính tiền
def recognize_and_calculate_bill(cropped_images, model, class_indices, menu):
    total_bill = 0
    results = []

    for img_path in cropped_images:
        img = image.load_img(img_path, target_size=(224, 224))
        img_array = image.img_to_array(img)
        img_array = np.expand_dims(img_array, axis=0) / 255.0

        pred = model.predict(img_array)
        class_idx = np.argmax(pred[0])
        food_name = list(class_indices.keys())[class_idx]

        food_price = next((item['price'] for item in menu if item['name'] == food_name), 0)
        results.append((food_name, food_price))
        total_bill += food_price

    return results, total_bill

```

```python
# 10. Hàm test ảnh
def test_image():
    if not os.path.exists(model_path):
        raise FileNotFoundError(f"Không tìm thấy mô hình tại {model_path}. Vui lòng huấn luyện mô hình trước!")

    model = load_model(model_path)

    # Lấy class_indices từ dataset
    datagen = create_data_generator()
    train_generator = datagen.flow_from_directory(
        dataset_path,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical',
        subset='training'
    )
    class_indices = train_generator.class_indices

    # Đọc menu
    with open(menu_path, 'r') as f:
        menu = json.load(f)

    # Cắt ảnh bằng YOLO
    cropped_images = crop_foods(input_image_path, cropped_path)

    # Kiểm tra số món ăn
    if len(cropped_images) < 1:  # Điều chỉnh theo yêu cầu thực tế
        print("Lỗi: Không phát hiện được món ăn nào!")
        return

    # Nhận diện và tính tiền
    results, total_bill = recognize_and_calculate_bill(cropped_images, model, class_indices, menu)

    # In kết quả
    print("\nKết quả nhận diện:")
    for food_name, price in results:
        print(f"Món: {food_name}, Giá: {price} VNĐ")
    print(f"Tổng hóa đơn: {total_bill} VNĐ")

# 11. Main function
def main():
    try:
        if not os.path.exists(model_path):
            print("Huấn luyện mô hình...")
            model, class_indices = train_cnn_model()
        else:
            print("Tải mô hình đã huấn luyện...")
        test_image()
    except Exception as e:
        print(f"Lỗi: {str(e)}")

if __name__ == "__main__":
    main()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam

# Thông số
img_size = (128, 128)
batch_size = 32
epochs = 15
data_path = '/content/drive/MyDrive/Data AI'  # Thư mục chứa ảnh các món ăn

# Tiền xử lý dữ liệu
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_gen = datagen.flow_from_directory(
    data_path,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='categorical',
    subset='training'
)

val_gen = datagen.flow_from_directory(
    data_path,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='categorical',
    subset='validation'
)

# Mô hình CNN
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=img_size + (3,)),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Conv2D(64, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(train_gen.num_classes, activation='softmax')
])

model.compile(optimizer=Adam(0.0005), loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(train_gen, validation_data=val_gen, epochs=epochs)

# Lưu mô hình
model.save('/content/cnn_monan_model.h5')

# Lưu label
label_map = train_gen.class_indices
inverse_labels = {v: k for k, v in label_map.items()}
import json
with open('/content/label_map.json', 'w') as f:
    json.dump(inverse_labels, f)
```
